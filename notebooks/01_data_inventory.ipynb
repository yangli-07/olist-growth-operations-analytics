```python
from pathlib import Path
import pandas as pd

project_root = Path.cwd().parent

raw_dir = project_root / "data" / "raw"
reports_dir = project_root /"reports"

reports_dir.mkdir(exist_ok=True)

print("Project root:", project_root)
print("Raw data directory:", raw_dir)
```

```python
csv_files = sorted(raw_dir.glob("*.csv"))

print("Number of CSV files:", len(csv_files))

for file in csv_files:
    print(file.name)
```

```python
assert len(csv_files) == 9, f"Expected 9 CSV files, found {len(csv_files)}"
```

```python
inventory = []

for file in csv_files:
    df = pd.read_csv(file, low_memory=False)
    inventory.append({
        "file_name": file.name,
        "rows": len(df),
        "columns": len(df.columns),
        "duplicate_rows": int(df.duplicated().sum()),
        "missing_cells": int(df.isna().sum().sum()),
        "file_size_mb": round(file.stat().st_size / (1024 ** 2), 2)
    })

inventory_df = pd.DataFrame(inventory)

inventory_df
```

```python
output_file = reports_dir / "data_inventory.csv"

inventory_df.to_csv(output_file, index=False)

print("Inventory saved to:", output_file)
```
